```python
import pandas as pd

# --------------------------------------------------
# 1. Load datasets
# --------------------------------------------------

ff = pd.read_csv(
    "ff_factors/europe_ff5_mom_daily.csv",
    parse_dates=["date"]
)

stoxx = pd.read_parquet(
    "stoxx600_daily_returns_2010_2025.parquet"
)

# --------------------------------------------------
# 2. Clean column names
# --------------------------------------------------

stoxx.columns = stoxx.columns.astype(str).str.strip()
ff.columns = ff.columns.astype(str).str.strip()

print("STOXX columns:")
print(stoxx.columns.tolist())

print("\nFama-French columns:")
print(ff.columns.tolist())

# --------------------------------------------------
# 3. Find the STOXX date column
# --------------------------------------------------

if "date" in stoxx.columns:
    stoxx_date_col = "date"
elif "datadate" in stoxx.columns:
    stoxx_date_col = "datadate"
else:
    raise KeyError(
        "Could not find a STOXX date column. "
        f"Available columns are: {stoxx.columns.tolist()}"
    )

print("\nUsing STOXX date column:", stoxx_date_col)

# --------------------------------------------------
# 4. Convert date columns
# --------------------------------------------------

stoxx[stoxx_date_col] = pd.to_datetime(stoxx[stoxx_date_col])
ff["date"] = pd.to_datetime(ff["date"])

# Rename STOXX date column to 'date'
stoxx = stoxx.rename(columns={stoxx_date_col: "date"})

# --------------------------------------------------
# 5. Merge on date
# --------------------------------------------------

merged = stoxx.merge(
    ff,
    on="date",
    how="inner"
)

# --------------------------------------------------
# 6. Check result
# --------------------------------------------------

print("\nSTOXX shape:", stoxx.shape)
print("Fama-French shape:", ff.shape)
print("Merged shape:", merged.shape)

print("\nMerged date range:")
print(merged["date"].min(), "to", merged["date"].max())

print("\nMerged columns:")
print(merged.columns.tolist())
merged.to_parquet("stoxx_600_returns_with_ff5_mom_daily_2010_2025.parquet", index=False)

display(merged.head())
```

```python
#there is a currency mismatch between ff factors (dollars) and returns, checking the returns currency
# Count observations by return currency
currency_counts = (
    merged["curcdd"]
    .value_counts(dropna=False)
    .reset_index()
)

currency_counts.columns = ["currency", "n_observations"]

currency_counts["share"] = (
    currency_counts["n_observations"] / len(merged)
)

display(currency_counts)
```
